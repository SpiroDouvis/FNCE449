### Connecting to the RIT case:

```python
import requests # step 1
API_KEY = {'X-API-key': 'FLI7E73K'} # step 2
def main():
    with requests.Session() as s: # step 3
        s.headers.update(API_KEY) # step 4
        resp = s.get('http://localhost:10002/v1/case') # step 5
        if resp.ok: 
            case = resp.json() 
            tick = case['tick'] # accessing the 'tick' value that was returned
            print('The case is on tick', tick) 
if __name__ == '__main__':
    main()
```

## Submitting orders to RIT

### Market orders:

```python
import requests # step 1
API_KEY = {'X-API-key': 'FLI7E73K'} # step 2
with requests.Session() as s: # step 3
    s.headers.update(API_KEY) # step 4
    mkt_buy_params={'ticker':'CRZY_M', 'type':'MARKET', 'quantity':1000, 'action':'BUY'}

    resp = s.post('http://localhost:9999/v1/orders',params=mkt_buy_params) # step 5
    if resp.ok: 
        mkt_order = resp.json() 
        id = mkt_order['order_id']
        print('The market buy order was submitted and has ID', id)
    else:
        print('The order was not successfully submitted!')
```
